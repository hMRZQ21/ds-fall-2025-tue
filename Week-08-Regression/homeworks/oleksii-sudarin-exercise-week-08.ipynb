# Rough Guide a v1 model.
0. Make a model to predict PTS 
0. For v1, dont use `2P, 3P, FG, or FT`.  Instead use `2PA, 3PA, FTA`.  
0. Include any other cols use see fit.  


````python
dependent_variable = 'PTS'
independent_variables = ['2PA', '3PA', 'FTA']

X = df[independent_variables]
y = df[dependent_variable]

X = sm.add_constant(X)

model_points = sm.OLS(y, X).fit()

y_pred = model_points.predict(X)
results = model_points.summary2()
print(results)
````

````python
test_df = pd.read_csv('../data/NBA_test.csv')

X_test = test_df[independent_variables]
y_test = test_df[dependent_variable]

X_test = sm.add_constant(X_test)

y_pred = round(model_points.predict(X_test),4)
````

````python
r_sq = round(metrics.r2_score(y_test, y_pred), 4)

model_root_mean_sq_error = round(rmse(y_test, y_pred), 4)

print('R^2 Score:', r_sq)
print('RMSE:', model_root_mean_sq_error)
````

````python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=y_test, y=y_pred)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='red')
plt.xlabel('Actual Points (y_test)')
plt.ylabel('Predicted Points (y_pred)')
plt.title('Actual vs. Predicted Points')
plt.grid(True)
plt.show()
````
