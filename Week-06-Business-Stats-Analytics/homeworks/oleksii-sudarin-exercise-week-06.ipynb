## Dataset Setup

First, let's generate the synthetic dataset you'll be analyzing:

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set random seed for reproducible results
np.random.seed(42)

# Generate 500 orders from 2023
n_orders = 500

# Create order IDs and customer data
order_id = np.arange(1, n_orders + 1)
customer_id = np.random.randint(1, 101, size=n_orders)
customer_segment = np.random.choice(["Student", "Professional", "Senior"], size=n_orders, p=[0.4, 0.4, 0.2])
region = np.random.choice(["North", "South", "East", "West"], size=n_orders)
order_date = pd.date_range(start="2023-01-01", periods=n_orders, freq="D")
product_category = np.random.choice(["Electronics", "Clothing", "Books", "Home", "Sports"], size=n_orders)
quantity = np.random.randint(1, 5, size=n_orders)
unit_price = np.round(np.random.uniform(10, 200, size=n_orders), 2)
revenue = quantity * unit_price
discount_applied = np.random.choice([0, 5, 10, 15, 20, 25, 30], size=n_orders, p=[0.5,0.1,0.1,0.1,0.05,0.05,0.1])
payment_method = np.random.choice(["Credit Card", "PayPal", "Gift Card", "Debit Card"], size=n_orders)

# Build dataframe
df = pd.DataFrame({
    "order_id": order_id,
    "customer_id": customer_id,
    "customer_segment": customer_segment,
    "region": region,
    "order_date": order_date,
    "product_category": product_category,
    "quantity": quantity,
    "unit_price": unit_price,
    "revenue": revenue,
    "discount_applied": discount_applied,
    "payment_method": payment_method
})

# Save dataset for reference
df.to_csv("ecommerce_data.csv", index=False)
print("✅ Dataset created successfully!")

# Display first few rows
df.head()
```

## Data Exploration

Let's start by understanding our dataset:

```python
# Load the data
data = pd.read_csv("ecommerce_data.csv")

# Basic information about the dataset
print("Dataset Info:")
print(data.info())
print("\nDataset Shape:", data.shape)
print("\nFirst few rows:")
data.head()
```

```python
# Convert order_date to datetime and add helpful columns
data["order_date"] = pd.to_datetime(data["order_date"])
data["month"] = data["order_date"].dt.to_period("M")
data["day_of_week"] = data["order_date"].dt.day_name()

# Example: Revenue by month
monthly_revenue = data.groupby("month")["revenue"].sum()
print("Monthly Revenue:")
print(monthly_revenue)

# Plot monthly revenue
monthly_revenue.plot(kind="bar", figsize=(10,6), title="Monthly Revenue")
plt.ylabel("Revenue ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### 1. Revenue Overview
What was the company's total revenue last quarter? Break it down by month.

```python
# Your code here
latest_quarter = data['order_date'].dt.to_period('Q').max()-1

monthly_revenue_last_quarter = data[data['order_date'].dt.to_period('Q') == latest_quarter].groupby("month")["revenue"].sum()
print(monthly_revenue_last_quarter)
monthly_revenue_last_quarter.plot(kind="bar", figsize=(10,6), title="Monthly Revenue Last Quarter")
plt.ylabel("Revenue ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 2. Top Customers
Who are the top 5 customers by total revenue contribution?

```python
# Your code here
top_customers = data.groupby('customer_id')['revenue'].sum().nlargest(5)
print(top_customers)

top_customers.plot(kind="barh", figsize=(10,6), title="Top 5 Customers")
plt.xlabel("Revenue ($)")
plt.ylabel('Customer ID')
plt.tight_layout()
plt.show()
```

### 3. Segment Analysis
Which customer segment (Student, Professional, Senior) spends the most on average per order?

```python
# Your code here
most_spending_segment = data.groupby('customer_segment')['revenue'].mean().sort_values()
print(most_spending_segment)

most_spending_segment.plot(kind="barh", figsize=(10,6), title="Most Spending Customer Segment")
plt.xlabel("Average Cost per Order ($)")
plt.ylabel('Customer Segment')
plt.tight_layout()
plt.show()
```

### 4. Regional Trends
Which region generated the highest revenue? Which region had the highest average order size?

```python
# Your code here
top_regions = data.groupby('region')['revenue'].agg(['sum', 'mean']).sort_values(by='sum')
print(top_regions)
fig, ax = plt.subplots(figsize=(10, 6))
top_regions['sum'].plot(kind='bar', ax=ax, label='Total Revenue (Sum)', rot=0, title='Revenue Metrics by Region: Total vs. Average')
ax.set_ylabel('Total Revenue ($)')
ax.tick_params(axis='y')
ax2 = ax.twinx()
top_regions['mean'].plot(kind='line', ax=ax2, marker='o', label='Average Order Size (Mean)', color='tab:red')
ax2.set_ylabel('Average Order Size ($)')
ax2.tick_params(axis='y')
plt.show()
```

### 5. Product Category Performance
Which product category is most popular by quantity vs. by revenue?

```python
# Your code here
popular_product_category = data.groupby('product_category')['revenue'].agg(['count', 'sum']).sort_values(by='sum')
print(popular_product_category)
fig, ax = plt.subplots(figsize=(10, 6))
popular_product_category['sum'].plot(kind='bar', ax=ax, label='Total Revenue (Sum)', rot=0, title='Product Quantity VS. Revenue')
ax.set_xlabel('Product Category')
ax.set_ylabel('Total Revenue ($)')
ax.tick_params(axis='y')
ax2 = ax.twinx()
popular_product_category['count'].plot(kind='line', ax=ax2, marker='o', label='Order Quantity (Count)', color='tab:red')
ax2.set_ylabel('Order Quantity')
ax2.tick_params(axis='y')
plt.show()
```

### 6. Discount Effectiveness
Do discounted orders generate higher or lower revenue per order compared to non-discounted orders?

```python
# Your code here
discount_effectiveness = data.groupby('discount_applied')['revenue'].mean()
print(discount_effectiveness)
discount_effectiveness.plot(kind="bar", figsize=(10,6), title="Discount Effectiveness")
plt.ylabel("Average Revenue ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 7. Payment Method Usage
What percentage of orders use each payment method? Does any payment method correlate with higher spending?

```python
# Your code here
payment_method_percentages = data.groupby('payment_method')['revenue'].agg(['count', 'sum'])
print(payment_method_percentages)
payment_method_percentages.plot(kind="pie", subplots=True, figsize=(10,6), title="Common Payment Method", legend=False, autopct='%1.1f%%')
```

### 8. Seasonality Check
Plot revenue by day of week — are weekends busier than weekdays?

```python
# Your code here
revenue_by_day_of_week = data.groupby("day_of_week")["revenue"].sum().reindex(['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
print(revenue_by_day_of_week)
revenue_by_day_of_week.plot(kind="line", figsize=(10,6), title="Revenue by the Day of the Week")
plt.ylabel("Revenue ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
